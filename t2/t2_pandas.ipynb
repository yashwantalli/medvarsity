cleaning


```python
df_dup=df.dropna()
print(df_dup.isnull().sum())    
print(df_dup.shape)
```

```python
df_dup1=df.fillna(0)
print(df_dup1.isnull().sum())    
print(df_dup1.shape)
```

```python
df_dup2=df.copy()
df_dup2["Age"]=df_dup2["Age"].fillna(df_dup2["Age"].median())
print(df_dup2.isnull().sum())
print(df_dup2.shape)    
```

```python
df.columns
```

```python
df_dup3=df.rename(columns={"PassengerId":"PId","Survived":"Survived_Stat"})
```

```python
df_dup3.columns
```

```python
df_dup4=df.drop(columns=["Embarked","SibSp","Parch"])
print(df_dup4.columns)
```

```python
def grp(x):
    if x<18:
        return "Child"
    elif x<60:
        return "Adult"
    else:
        return "Senior"

df_dup5=df.copy()
df_dup5["Age_Group"]=df_dup4["Age"].apply(grp)
print(df_dup5.head())

print(df_dup5.shape)
```
